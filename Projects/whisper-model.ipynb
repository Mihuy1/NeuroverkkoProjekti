# Transcription Using Whisper Models

In this workflow, we utilize the `whisper` library to transcribe audio files in Finnish (`fi`) using three different pre-trained models: `tiny`, `small`, and `medium`. The transcription process involves the following steps:

1. **Model Loading**: 
    - The `tiny`, `small`, and `medium` models are loaded using the `whisper.load_model()` function.

2. **Audio Transcription**:
    - The models are used to transcribe two audio files: `test.m4a` and `testi2.m4a`.
    - The `transcribe()` method is called with the audio file and the specified language (`fi` for Finnish).

3. **Results**:
    - The transcriptions for each model are printed for both audio files, allowing a comparison of the outputs from the different models.

This process demonstrates the capabilities of the Whisper models in handling multilingual transcription tasks and provides insights into the performance of different model sizes.

```python
import whisper
import numpy as np

tinyModel = whisper.load_model('tiny')
smallModel = whisper.load_model('small')
mediumModel = whisper.load_model('medium')


resultTiny = tinyModel.transcribe("test.m4a", language="fi")
resultSmall = smallModel.transcribe("test.m4a", language="fi")
resultMedium = mediumModel.transcribe("test.m4a", language="fi")

print("Tiny Model Transcription:", resultTiny["text"])
print("Small Model Transcription:", resultSmall["text"])
print("Medium Model Transcription:", resultMedium["text"])

resultTiny = tinyModel.transcribe("testi2.m4a", language="fi")
resultSmall = smallModel.transcribe("testi2.m4a", language="fi")
resultMedium = mediumModel.transcribe("testi2.m4a", language="fi")

print("Another Tiny Model Transcription:", resultTiny["text"])
print("Another Small Model Transcription:", resultSmall["text"])
print("Another Medium Model Transcription:", resultMedium["text"])


```
