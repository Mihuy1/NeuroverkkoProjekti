```python
import whisper
import numpy as np

tinyModel = whisper.load_model('tiny')
smallModel = whisper.load_model('small')
mediumModel = whisper.load_model('medium')


resultTiny = tinyModel.transcribe("test.m4a", language="fi")
resultSmall = smallModel.transcribe("test.m4a", language="fi")
resultMedium = mediumModel.transcribe("test.m4a", language="fi")

print("Tiny Model Transcription:", resultTiny["text"])
print("Small Model Transcription:", resultSmall["text"])
print("Medium Model Transcription:", resultMedium["text"])

resultTiny = tinyModel.transcribe("testi2.m4a", language="fi")
resultSmall = smallModel.transcribe("testi2.m4a", language="fi")
resultMedium = mediumModel.transcribe("testi2.m4a", language="fi")

print("Another Tiny Model Transcription:", resultTiny["text"])
print("Another Small Model Transcription:", resultSmall["text"])
print("Another Medium Model Transcription:", resultMedium["text"])


```
